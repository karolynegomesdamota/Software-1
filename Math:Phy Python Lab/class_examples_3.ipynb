```python
x = np.arange(-5, 5, 0.001)
y1 = x**2 -10
y2 = x * 4
y = (x < 0) * y1 + (x >= 0) * y2 # pi

fig, ax = plt.subplots(3, figsize = (10,9), sharex = True, sharey=True)
plt.sca (ax[0])
plt.plot(x,y1, color = "b", label = "y1")
plt.grid(True)
plt.legend(loc = "lower left")

plt.sca (ax[1])
plt.plot(x,y2, color = "r", label = "y2")
plt.grid(True)
plt.legend(loc = "lower right")

plt.sca (ax[2])
plt.plot(x,y, color = "y", label = "x")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend(loc = "upper left")
plt.show()
```
