### My solution

```python
# start your code here

x = np.linspace(-10, 10, 2000)

y1 = 0
y2 = 0.2 + x
y3 = 5.8 - x

y = (x < -0.2) * y1 + ((-0.2 <= x) & (x < 2.8)) * y2 + ((2.8 <= x) & (x < 5.8)) * y3 + (x >= 5.8) * y1

plt.figure(figsize = (8, 5))
plt.plot(x, y, label='f(x)')
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(f"Problem 1: Piecewise Function")
plt.legend()
plt.grid(True)
plt.show()
```
