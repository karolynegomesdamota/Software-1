## My solution for problem 1

```python
# Your code starts here

# Create an overview graph

x = np.linspace(-10, 10, 1000)

# Find all roots graphically

# A root is where the function crosses the x-axis (where f(x) = 0, meaning y = 0)

y = mx.lab_2_problem_1(x, student_id)

plt.figure(figsize = (8, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(f"All roots visible")
plt.grid(True)
plt.xlim(-10, 10) # X Limit, meaning we will zoom in this range of X
plt.ylim(-1, 1) # Y Limit, meaning we will zoom in this range of Y
plt.plot(-4.1501, 0, "ro") # To mark the exact point # ro = round
plt.plot(-2.9656, 0, "ro") # To mark the exact point # ro = round
plt.plot(6.0459, 0, "ro") # To mark the exact point # ro = round
plt.plot(7.4727, 0, "ro") # To mark the exact point # ro = round
plt.axhline(y=0, color='red', linestyle='--')
plt.show()

plt.figure(figsize = (8, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("First root (zoomed)")
plt.grid(True)
plt.xlim(-4.2, -4.1) # X Limit, meaning we will zoom in this range of X
plt.ylim(-1, 1) # Y Limit, meaning we will zoom in this range of Y
plt.plot(-4.1501, 0, "ro") # To mark the exact point # ro = round
plt.text (-4.14, 0, "$x\\approx -4.1501$", color="red")
plt.show()

plt.figure(figsize = (8, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Second root (zoomed)")
plt.grid(True)
plt.xlim(-3, -2.9) # X Limit, meaning we will zoom in this range of X
plt.ylim(-1, 1) # Y Limit, meaning we will zoom in this range of Y
plt.plot(-2.9656, 0, "ro") # To mark the exact point # ro = round
plt.text (-2.96, 0, "$x\\approx -2.9656$", color="red")
plt.show()

plt.figure(figsize = (8, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Third root (zoomed)")
plt.grid(True)
plt.xlim(6, 6.1) # X Limit, meaning we will zoom in this range of X
plt.ylim(-1, 1) # Y Limit, meaning we will zoom in this range of Y
plt.plot(6.0459, 0, "ro") # To mark the exact point # ro = round
plt.text (6.05, 0, "$x\\approx 6.0459$", color="red")
plt.show()

plt.figure(figsize = (8, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Forth root (zoomed)")
plt.grid(True)
plt.xlim(7.4, 7.5) # X Limit, meaning we will zoom in this range of X
plt.ylim(-1, 1) # Y Limit, meaning we will zoom in this range of Y
plt.plot(7.4727, 0, "ro") # To mark the exact point # ro = round
plt.text (7.48, 0, "$x\\approx 7.4727$", color="red")
plt.show()

```

```python
# Your code starts here

x = np.linspace(0, 10, 1000)
fx = np.sqrt(4.4*x)
gx = (x - 6.51)**2

plt.plot(x, fx, label = r"$ f(x) = \sqrt{4.4x} $")
plt.plot(x, gx, label = "$ g(x) = (x - 6.51)^2 $")
plt.xlim(0, 10) # X Limit, meaning we will zoom in this range of X
plt.ylim(-1, 20) # Y Limit, meaning we will zoom in this range of Y
plt.grid(True)
plt.title("Both function in the same graph")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()

plt.plot(x, fx, label = r"$ f(x) = \sqrt{4.4x} $")
plt.plot(x, gx, label = "$ g(x) = (x - 6.51)^2 $")
plt.xlim(4, 5) # X Limit, meaning we will zoom in this range of X
plt.ylim(0, 10) # Y Limit, meaning we will zoom in this range of Y
plt.plot(4.41, 4.41, "ro") # To mark the exact point # ro = round
plt.text (4.5, 4.41, "$approx (4.41, 4.41)$", color="red")
plt.grid(True)
plt.title("Intersection 1")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()

plt.plot(x, fx, label = r"$ f(x) = \sqrt{4.4x} $")
plt.plot(x, gx, label = "$ g(x) = (x - 6.51)^2 $")
plt.xlim(8, 10) # X Limit, meaning we will zoom in this range of X
plt.ylim(0, 10) # Y Limit, meaning we will zoom in this range of Y
plt.plot(9, 6.3, "ro") # To mark the exact point # ro = round
plt.text (9.1, 6.3, "$approx (9, 6.3)$", color="red")
plt.grid(True)
plt.title("Intersection 2")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```
