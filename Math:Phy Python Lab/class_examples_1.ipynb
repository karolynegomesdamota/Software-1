```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create and convert a list into an array

a = [1, 2, 3, 4]
x = np.array(a)
print(type(x))

y = np.array([1, 2, 3, 4])

# Create an array using range

z = np.arange(7)
print(z)

z1 = np.arange(-2, 4)
print(z1)

z12 = np.arange(-2, 4, 0.05)
print(z12)
```

```python
# Generate an array of points

x = np.linspace(0, 5, 200)

print(x)
```

```python
# Square roots

x = np.array([0, 4, 25, 16])
y = np.sqrt(x)

print(x)
print(y)
```

```python
# Different way to use arrays:

x = np.linspace(-2, 2, 9)
y = 2*x + 1

print(x)
print(y)
```

```python
x = np.arange(0, 4, 0.05)
y = np.sin(x * np.pi)

# Object-oriented approach

fig, ax = plt.subplots(figsize = (6, 4))
ax.plot(x, y)
ax.set_xlabel("Time [s]")
ax.set_ylabel("Amplitude [V]")
ax.set_title("Sine wave")
fig.set_facecolor("lightblue")
```

```python
# Pyplot approach:

x = np.arange(0, 4, 0.05)
y = np.sin(x * np.pi)

plt.figure(figsize = (8, 5))
plt.plot(x, y)
plt.xlabel("Time [s]")
plt.ylabel("Amplitude [V]")
plt.title("Sine wave")
plt.grid(True)
plt.axhline(0.0, color = "red")
plt.axvline(2.0, color = "green")
```

```python
x = np.linspace(-5, 5, 200)
y = x**2 -2*x -5
plt.figure(figsize=(8,5))
plt.plot(x, y, label="f(x)")
plt.axhline(0, color="red", linestyle="--", label="$y=0$")
plt.title("Graphical solution of the equation $f(x) = 0$")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.xlim(-1.5, -1.4) # zoom in the exact point
plt.ylim(-0.5, 0.5) # zoom in the exact point
plt.scatter(-1,45, 9, color = "red") #mark specific point
plt.text (-1.44, 0.05, "$x\\approx - 1.45$", color="red") #using latex syntax
plt.legend()
plt.grid(True)
plt.show()
```
