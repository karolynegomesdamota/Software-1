## My solution for problem 2

```python
# An intersection point is a value of x where f(x) = c. Graphically, this is where the function crosses the horizontal line where y = c.

# In order to clearly see each point, I had break it down into 2 graphs:

# First point

plt.figure(figsize = (7, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(f"Function f(x) for student {student_id} - Zoomed view for first point")
plt.grid(True)

plt.xlim(5.5, 5.55) # zoom in the exact point
plt.ylim(0.5, 1.5) # zoom in the exact point
plt.text (5.53, 1, "$x\\approx 5.53$", color="red")
plt.axhline(y=1, color='red', linestyle='--')
plt.grid(True)
plt.show()

# Second point

plt.figure(figsize = (7, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(f"Function f(x) for student {student_id} - Zoomed view for second point")
plt.grid(True)

plt.xlim(7.95, 8) # zoom in the exact point
plt.ylim(0.5, 1.5) # zoom in the exact point
plt.text (7.99, 1, "$x\\approx 7.99$", color="red")
plt.axhline(y=1, color='red', linestyle='--')
plt.grid(True)
plt.show()

# Zoomed in graph with both points

plt.figure(figsize = (7, 5))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title(f"Function f(x) for student {student_id} - Zoomed view")
plt.grid(True)

plt.xlim(5, 9) # zoom in the exact point
plt.ylim(0.5, 1.5) # zoom in the exact point
plt.text (5.53, 1, "$x\\approx 5.53$", color="red")
plt.text (7.99, 1, "$x\\approx 7.99$", color="red")
plt.axhline(y=1, color='red', linestyle='--')
plt.grid(True)
plt.show()

```
