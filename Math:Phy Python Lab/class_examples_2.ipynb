```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a 4th order polynomial function
x = np.linspace(0,4)
y = 0.1 * x ** 4 - 0.5 * x ** 3 + 0.2 * x ** 2 - 0.1 * x + 4

fig, ax = plt.subplots(figsize = (10, 6))
plt.plot(x, y, linestyle = "--", linewidth = 5, label = "4th order polynomial function")
plt.grid(True)
plt.xlabel("x", fontsize = 18)
plt.ylabel("y", fontsize = 18)
plt.title ("Graph of a function", fontsize = 21)
plt.legend()
plt.show()
```

```python
# Create multiple subplots figure

fig, ax = plt.subplots(2, figsize = (10, 8))
x = np.linspace(0, 5)
y1 = x - 3
plt.sca(ax[0]) # sca(): set the current area
plt.plot(x, y1, color = "yellow", label = "x - 3")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
# Do not include show here since it will cause an error. Call show only at the end.

y2 = x**2 - 4*x
plt.sca(ax[1])
plt.plot(x, y2, color = "blue", label = "$x^2 -4x$") # Writing maths expression on a figure using latex syntax
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
# Create multiple figures

fig1, ax1 = plt.subplots()
x = np.linspace(0, 5)
y1 = x - 3
plt.plot(x, y1, color = "yellow", label = "x - 3")
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
# Do not include show here since it will cause an error. Call show only at the end.

fig2, ax2 = plt.subplots()
y2 = x**2 - 4*x
plt.plot(x, y2, color = "blue", label = "$x^2 -4x$") # Writing maths expression on a figure using latex syntax
plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
fig, ax = plt.subplots()
x = np.linspace(0, 5)
y1 = x - 3
y2 = x**2 - 4*x

plt.plot(x, y1, label = "$ y = x -3 $")
plt.plot(x, y2, label = "$ y = x^2 -4x$")

plt.grid(True)
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
# Pyplot approach to work directly with a figure

x = np.linspace(-10, 10, 200) # Along the way we changed 100 values to 200 to make the second graph smoother
y = x ** 3 - 2.0 * x ** 2 - 1.0 * x + 2.0

plt.figure(figsize = (8, 8))
plt.subplot(2, 1, 1) # Number of subplots / Row / Column
plt.title("Original function")
plt.plot(x, y)
plt.grid(True)

plt.subplot(2, 1, 2)
plt.plot(x, y)
plt.xlim(-5, 5) # X Limit, meaning we will zoom in this range of X
plt.ylim(-2, 3) # Y Limit, meaning we will zoom in this range of Y
plt.plot(0, 2, "ro") # To mark the exact point # ro = round
plt.title("Customized axes-limits")
plt.grid(True)

plt.show()
```
